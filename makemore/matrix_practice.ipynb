# Tensor Practice: Slicing, Multiplication, Broadcasting & Ops

Hands-on drills for the core PyTorch tensor moves used throughout this repo
(see `broadcasting_tutorial.ipynb` and `makemore_MLP.ipynb`).

**How to use this notebook:**
1. Run the setup cell for an exercise (creates the input tensors).
2. Fill in `result = ...` in the same cell, replacing the `None`.
3. Run the **check cell** right after it. It tells you whether your `result`
   has the right **shape**, and — only if the shape is right — whether the
   **values** are correct. It never prints or reveals the correct answer, so
   you can retry without being spoiled.
4. If you get stuck, the **Solutions** section at the very bottom has the
   answer for every exercise — but no peeking until you've had a real go!

Run the next cell once before starting.

```python
import torch
import hashlib

def _tensor_hash(t):
    t = torch.as_tensor(t).detach().cpu().contiguous().double()
    rounded = torch.round(t * 1e4) / 1e4
    payload = rounded.numpy().tobytes() + str(tuple(rounded.shape)).encode()
    return hashlib.sha256(payload).hexdigest()[:16]

def check(name, result, expected_shape, expected_hash):
    """Verify `result` against a hidden expected shape/hash without revealing it."""
    if result is None:
        print(f"[{name}] -> fill in `result` above, then re-run this cell.")
        return
    result = torch.as_tensor(result)
    got_shape = tuple(result.shape)
    if got_shape != expected_shape:
        print(f"[{name}] shape mismatch: got {got_shape}, expected {expected_shape}")
        return
    if _tensor_hash(result) == expected_hash:
        print(f"[{name}] correct!")
    else:
        print(f"[{name}] shape is right, but the values are off. Keep trying.")
```
